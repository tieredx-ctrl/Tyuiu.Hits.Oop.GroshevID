Объектно-ориентированное программирование (ООП) в C# включает в себя несколько ключевых концепций, которые помогают структурировать программы и улучшить их читаемость, поддерживаемость и расширяемость.

<h2 style="color:DodgerBlue">1. Классы и Объекты</h2>
Классы в C# - это шаблоны для создания объектов. Они определяют набор свойств (данных) и методов (действий), которые могут выполнять эти объекты. Объекты создаются по этим шаблонам и представляют собой экземпляры классов.

<h3 style="color:DodgerBlue">Пример кода:</h3>


```csharp
// Создание класса
class Vehicle
{

}

// Основная программа
Vehicle Car = new Vehicle(); // Создание объекта
```

### Разбор: классы и объекты

**Теория.** Класс — описание того, какие данные и действия есть у объекта. Объект — конкретный экземпляр класса. Например, Animal описывает животное, а animal — конкретное животное по имени Шарик. Несколько объектов одного класса могут иметь разные имена.

**Что происходит в решении:**
1. `class Animal` объявляет класс; это ещё не создание животного.
2. Поле `Name` хранит имя. `public` разрешает обращаться к нему извне.
3. Метод `ShowInfo()` выводит имя. `void` означает, что метод не возвращает значение.
4. `new Animal()` создаёт объект; переменная `animal` хранит ссылку на него.
5. `animal.Name = "Шарик"` меняет имя именно этого объекта.
6. `animal.ShowInfo()` вызывает действие объекта и выводит `Животное: Шарик`.


<h2 style="color:DodgerBlue">2. Инкапсуляция</h2>

Инкапсуляция - это процесс скрытия деталей реализации класса от пользователя. В C#, инкапсуляция достигается за счет использования модификаторов доступа (`public`, `private`, `protected`), а также свойства и методы для управления доступом к данным объекта.

```csharp
// Создание класса
class Vehicle
{
    private string color; // Приватное поле

    public string Color // Свойство с геттером и сеттером
    {
        get { return color; }
        set { color = value; }
    }

    public void StartEngine() // Публичный метод
    {
        Console.WriteLine("Запустить двигатель...");
    }
}

// Основная программа
Vehicle Car = new Vehicle(); // Создание объекта
Car.Color = "Красный";
Car.StartEngine();
Console.WriteLine(Car.Color);

```

<h4 style="color:DodgerBlue">Для проверки напишите пример кода ниже в блоке:</h4>

----

````polyglot-notebook
// Создание класса
class Vehicle
{
    private string color; // Приватное поле

    public string Color // Свойство с геттером и сеттером
    {
        get { return color; }
        set { color = value; }
    }

    public void StartEngine() // Публичный метод
    {
        Console.WriteLine("Запустить двигатель...");
    }
}

// Основная программа
Vehicle Car = new Vehicle(); // Создание объекта
Car.Color = "Красный";
Car.StartEngine();
Console.WriteLine(Car.Color);
````

Ниже в блоке по примеру создайте класс Animal и реализуйте инкапсуляцию 

````polyglot-notebook
class Animal
{
    // Поле нельзя читать или изменять напрямую извне класса.
    private string name = "Без имени";

    public string Name
    {
        get { return name; }
        set
        {
            // Пустое имя не принимается: прежнее значение сохраняется.
            if (!string.IsNullOrWhiteSpace(value))
                name = value;
        }
    }

    public void Eat()
    {
        Console.WriteLine($"{name} ест.");
    }
}

Animal encapsulatedAnimal = new Animal();
encapsulatedAnimal.Name = "Шарик";
Console.WriteLine($"Имя: {encapsulatedAnimal.Name}");
encapsulatedAnimal.Eat();

encapsulatedAnimal.Name = "";
Console.WriteLine($"После попытки задать пустое имя: {encapsulatedAnimal.Name}");
````

### Разбор: инкапсуляция

**Теория.** Инкапсуляция объединяет данные и действия в классе и ограничивает доступ к внутренним деталям. `private` доступен внутри класса, `public` — извне, `protected` — также в наследниках. Поле хранит значение; свойство задаёт правила его чтения и изменения.

**Что происходит в решении:**
1. `private string name` скрывает поле: обратиться к `encapsulatedAnimal.name` снаружи нельзя.
2. Публичное свойство `Name` предоставляет разрешённый доступ. `get` возвращает имя, `set` получает новое значение в переменной `value`.
3. `string.IsNullOrWhiteSpace(value)` проверяет, является ли значение null, пустой строкой или только пробелами. Знак `!` означает «не».
4. Имя Шарик принимается, затем `Eat()` выводит `Шарик ест.`.
5. Попытка присвоить пустое имя отклоняется условием в `set`; имя остаётся Шарик. Так класс защищает своё состояние.


<h2 style="color:DodgerBlue">3. Наследование</h2>

Наследование позволяет создавать новые классы на основе существующих, наследуя их свойства и методы. Это упрощает повторное использование кода и расширение функциональности.

```csharp

class Vehicle
{
    private string color; // Приватное поле

    public string Color // Свойство с геттером и сеттером
    {
        get { return color; }
        set { color = value; }
    }
    public void Move()
    {
        Console.WriteLine("Поехали...");
    }
}

class Car : Vehicle
{

}

Vehicle myVehicle = new Vehicle(); // Создание объекта myVehicle
Car myCar = new Car(); // Создание объекта myCar

myCar.Color = "Красный";
Console.WriteLine(myCar.Color);
myCar.Move();

myVehicle.Color = "Синий";
Console.WriteLine(myVehicle.Color);
myVehicle.Move();

```

<h4 style="color:DodgerBlue">Для проверки напишите пример кода ниже в блоке:</h4>

----

````polyglot-notebook
class Vehicle
{
    private string color; // Приватное поле

    public string Color // Свойство с геттером и сеттером
    {
        get { return color; }
        set { color = value; }
    }
    public void Move()
    {
        Console.WriteLine("Поехали...");
    }
}

class Car : Vehicle
{

}

Vehicle myVehicle = new Vehicle(); // Создание объекта myVehicle
Car myCar = new Car(); // Создание объекта myCar

myCar.Color = "Красный";
Console.WriteLine(myCar.Color);
myCar.Move();

myVehicle.Color = "Синий";
Console.WriteLine(myVehicle.Color);
myVehicle.Move();
````

Ниже в блоке по примеру создайте класс Animal и реализуйте наследование 

````polyglot-notebook
class Animal
{
    public string Name { get; set; }

    public void Move()
    {
        Console.WriteLine($"{Name} двигается.");
    }
}

// Собака является животным и наследует Name и Move().
class Dog : Animal
{
    public void Bark()
    {
        Console.WriteLine($"{Name} лает: Гав!");
    }
}

Animal ordinaryAnimal = new Animal();
ordinaryAnimal.Name = "Животное";
ordinaryAnimal.Move();

Dog inheritedDog = new Dog();
inheritedDog.Name = "Шарик";
inheritedDog.Move();
inheritedDog.Bark();
````

### Разбор: наследование

**Теория.** Наследование позволяет создавать более конкретный класс на основе общего. Его удобно читать как отношение «является»: собака является животным. Общие доступные свойства и методы не нужно писать заново.

**Что происходит в решении:**
1. Базовый `Animal` содержит свойство `Name` и метод `Move()`.
2. `class Dog : Animal` объявляет Dog наследником Animal; двоеточие обозначает наследование.
3. Dog получает `Name` и `Move()` и добавляет собственный метод `Bark()`.
4. Сначала создаётся обычное животное, затем собака Шарик.
5. `inheritedDog.Move()` использует унаследованный метод, а `inheritedDog.Bark()` — метод собаки.

Вывод: `Животное двигается.`, `Шарик двигается.`, `Шарик лает: Гав!`. Private-члены базового класса наследник не может использовать напрямую.


<h2 style="color:DodgerBlue">4. Полиморфизм</h2>

Полиморфизм позволяет использовать один интерфейс для различных типов данных. В C# это достигается через переопределение методов и перегрузку методов и операторов.
Метод Move() может быть переопределен в классе Car для выполнения специфической для автомобилей логики:

```csharp
class Vehicle
{
    public virtual void Move()
    {
        Console.WriteLine("Транспорт двигается");
    }
}

class Car : Vehicle
{
    public override void Move()
    {
        //base.Move(); // Вызов метода базового класса
        Console.WriteLine("Автомобиль двигается");
    }
}

Vehicle myVehicle = new Vehicle(); // Создание объекта myVehicle
Car myCar = new Car(); // Создание объекта myCar


myVehicle.Move();
myCar.Move();
```

<h4 style="color:DodgerBlue">Для проверки напишите пример кода ниже в блоке:</h4>

----

````polyglot-notebook
class Vehicle
{
    public virtual void Move()
    {
        Console.WriteLine("Транспорт двигается");
    }
}

class Car : Vehicle
{
    public override void Move()
    {
        //base.Move(); // Вызов метода базового класса
        Console.WriteLine("Автомобиль двигается");
    }
}

Vehicle myVehicle = new Vehicle(); // Создание объекта myVehicle
Car myCar = new Car(); // Создание объекта myCar


myVehicle.Move();
myCar.Move();
````

Ниже в блоке по примеру создайте класс Animal и реализуйте полиморфизм 

````polyglot-notebook
class Animal
{
    // Наследники могут изменить реализацию этого метода.
    public virtual void MakeSound()
    {
        Console.WriteLine("Животное издаёт звук.");
    }
}

class Dog : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Гав!");
    }
}

class Cat : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Мяу!");
    }
}

// Один общий тип Animal, но разные реальные типы объектов.
Animal[] animals = { new Animal(), new Dog(), new Cat() };
foreach (Animal currentAnimal in animals)
    currentAnimal.MakeSound();
````

### Разбор: полиморфизм

**Теория.** Через общий тип можно работать с разными объектами, а одинаковый вызов метода даёт поведение конкретного объекта. В этом решении показан полиморфизм через переопределение. Перегрузка — отдельный приём: несколько методов с одним именем, но разными параметрами.

**Что происходит в решении:**
1. `virtual` разрешает наследникам изменить реализацию `MakeSound()`.
2. Dog и Cat пишут свои версии метода с `override`.
3. Массив `Animal[]` содержит обычное животное, собаку и кошку.
4. Цикл `foreach` вызывает одинаковую строку `currentAnimal.MakeSound()` для каждого объекта.
5. Выполняется версия, соответствующая реальному типу объекта: общий звук, `Гав!`, затем `Мяу!`.

Это не три разных имени метода, а один общий вызов с разным поведением.


<h2 style="color:DodgerBlue">5. Абстракция</h2>

Если бы у нас был класс ElectricCar, который должен иметь некоторые уникальные методы, мы могли бы определить абстрактный класс Car с абстрактными методами, обязательными для реализации в производных классах:

```csharp

abstract class Car
{
    public abstract void ChargeBattery();

    public void Move()
    {
        Console.WriteLine("Поехали...");
    }
}

class ElectricCar : Car
{
    public override void ChargeBattery()
    {
        Console.WriteLine("Батарея заряжена...");
    }
}

ElectricCar myElectricCar = new ElectricCar(); // Создание объекта myElectricCar

myElectricCar.ChargeBattery();
myElectricCar.Move();

```

<h4 style="color:DodgerBlue">Для проверки напишите пример кода ниже в блоке:</h4>

----

````polyglot-notebook
abstract class Car
{
    public abstract void ChargeBattery();

    public void Move()
    {
        Console.WriteLine("Поехали...");
    }
}

class ElectricCar : Car
{
    public override void ChargeBattery()
    {
        Console.WriteLine("Батарея заряжена...");
    }
}

ElectricCar myElectricCar = new ElectricCar(); // Создание объекта myElectricCar

myElectricCar.ChargeBattery();
myElectricCar.Move();
````

<h4 style="color:Red">Задание:</h4>

----

Ниже в блоке по примеру создайте класс Animal и реализуйте абстракцию 

````polyglot-notebook
// Animal задаёт общие действия, но не создаётся напрямую.
abstract class Animal
{
    // Тело метода должен написать конкретный наследник.
    public abstract void MakeSound();

    public void Eat()
    {
        Console.WriteLine("Животное ест.");
    }
}

class Dog : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Гав!");
    }
}

Animal abstractAnimal = new Dog();
abstractAnimal.MakeSound();
abstractAnimal.Eat();
````

### Разбор: абстракция

**Теория.** Абстракция выделяет существенные данные и действия, не требуя знать все детали реализации. Абстрактный класс — один из способов её выразить: он задаёт общее описание для конкретных наследников.

**Что происходит в решении:**
1. `abstract class Animal` запрещает создавать объект выражением `new Animal()`.
2. `public abstract void MakeSound();` объявляет действие без тела. Конкретный наследник обязан реализовать его.
3. `Eat()` уже имеет тело и доступен наследникам.
4. Dog реализует `MakeSound()` с помощью `override`.
5. В строке `Animal abstractAnimal = new Dog()` тип переменной общий, а созданный объект — собака.
6. Вызовы выводят `Гав!` и `Животное ест.`.

Отличие: virtual-метод уже имеет реализацию и разрешает её изменить; abstract-метод реализации не имеет и требует её написать.


<h2 style="color:DodgerBlue">6. Ассоциация (Композиция и Агрегация)</h2>

Композиция и агрегация являются двумя ключевыми концепциями в объектно-ориентированном программировании (ООП), включая C#. Они обе относятся к способам организации классов и их взаимодействия между собой, но имеют разные цели и способы реализации.

<h3 style="color:DodgerBlue">Композиция</h3>
Композиция — это форма ассоциации, которая описывает "часть-целое" отношение между классами. В контексте композиции один класс является частью другого класса, и они работают вместе для достижения общей цели. Классы, участвующие в композиции, тесно связаны друг с другом, и если одна часть (подкласс) уничтожается или изменяется, это может повлиять на работу всей системы.


<h3 style="color:DodgerBlue">Пример кода:</h3>

```csharp
using System;

public class Engine
{
    public void Start() => Console.WriteLine("Двигатель стартовал");
    public void Stop() => Console.WriteLine("Двигатель остановлен");
}

public class Car
{
    private Engine engine;

    public Car(Engine engine)
    {
        this.engine = engine;
    }

    public void Drive()
    {
        engine.Start();

        // Дополнительная логика для движения автомобиля
        Console.WriteLine("Автомобиль двигается");

        engine.Stop();
    }
}

Engine myEngine = new Engine(); // Создание объекта myEngine
Car myAuto = new Car(myEngine); // Создание объекта myAuto

myAuto.Drive(); // Автомобиль начинает движение, используя двигатель

```

<h4 style="color:DodgerBlue">Для проверки напишите пример кода ниже в блоке:</h4>

----

````polyglot-notebook
class Engine
{
    public void Start() { Console.WriteLine("Двигатель стартовал"); }
    public void Stop() { Console.WriteLine("Двигатель остановлен"); }
}

class Car
{
    private readonly Engine engine = new Engine();

    public void Drive()
    {
        engine.Start();
        Console.WriteLine("Автомобиль двигается");
        engine.Stop();
    }
}

Car composedCar = new Car();
composedCar.Drive();
````

<h4 style="color:Red">Задание:</h4>

----

Ниже в блоке по примеру создайте класс Animal и реализуйте композицию 

````polyglot-notebook
class Animal
{
    // Животное создаёт свою внутреннюю часть.
    // Ссылку на неё нельзя получить снаружи или заменить.
    private readonly Heart heart = new Heart();

    public void Live()
    {
        heart.Beat();
        Console.WriteLine("Животное живёт.");
    }

    // Внутренний тип доступен только внутри Animal.
    private class Heart
    {
        public void Beat()
        {
            Console.WriteLine("Сердце бьётся.");
        }
    }
}

Animal composedAnimal = new Animal();
composedAnimal.Live();
````

### Разбор: композиция

**Теория.** Композиция — отношение «содержит» с владением частью: в этой модели сердце принадлежит конкретному животному. Это отличается от наследования: животное содержит сердце, но не является сердцем.

**Что происходит в решении:**
1. Внутри Animal объявлен закрытый тип `Heart`. Другой код не может создать именно эту внутреннюю часть напрямую.
2. `new Heart()` выполняется при создании Animal: объект создаёт собственное сердце.
3. Поле `heart` закрыто (`private`), а `readonly` запрещает заменить ссылку после инициализации. Это не означает, что сам объект Heart обязательно неизменяемый.
4. `Live()` обращается к сердцу через `heart.Beat()`.
5. Вывод: `Сердце бьётся.`, затем `Животное живёт.`.

В исходном текстовом примере Engine передавался в Car извне; один этот факт не показывает владение частью. В заполненной ячейке примера Car создаёт свой Engine внутри. В C# нет команды, которая при композиции автоматически «уничтожает часть»: память освобождает сборщик мусора, когда объекты больше недоступны.


<h2 style="color:DodgerBlue">7. Ассоциация (Композиция и Агрегация)</h2>

<h3 style="color:DodgerBlue">Агрегация</h3>
Агрегация — это еще одна форма ассоциации, которая также описывает отношения между классами, но с более слабой связностью. В агрегации один класс содержит другой класс как составную часть, но эти классы могут функционировать независимо друг от друга. Если одна часть агрегата уничтожается, это не обязательно влияет на остальную систему.

Для реализации агрегации на примере классов `Car` (автомобиль) и `Driver` (водитель) с использованием абстрактного класса, мы можем определить базовый абстрактный класс `Vehicle` (Транспортное средство), который будет содержать общие характеристики для всех видов транспортных средств. Затем, мы определим классы `Car` и `Driver`, которые будут агрегироваться вместе, но будут независимы друг от друга.

### Шаги для реализации:

1. **Определение абстрактного класса `Vehicle`:** Этот класс будет служить базовым классом для всех транспортных средств, определяя общие свойства и методы.

2. **Определение класса `Driver`:** Этот класс будет содержать информацию о водителе и методы, связанные с его действиями за рулем.

3. **Определение класса `Car`:** Этот класс будет наследоваться от `Vehicle` и будет агрегировать `Driver`, но при этом будет независим от него.

<h3 style="color:DodgerBlue">Пример кода:</h3>

```csharp

using System;

abstract class Vehicle
{
    public abstract void Drive();
}

class Car : Vehicle
{
    private Driver driver;

    public Car(Driver driver)
    {
        this.driver = driver;
    }

    public override void Drive()
    {
        Console.WriteLine("Автомобиль движется...");
    }
}

class Driver
{
    public string Name { get; set; }

    public Driver(string name)
    {
        Name = name;
    }

    public void Drive(Vehicle vehicle)
    {
        Console.WriteLine($"{Name} управляет траспортом...");
        vehicle.Drive();
    }
}

    Driver Igor = new Driver("Игорь");
    Car myCar = new Car(Igor); // Создаем автомобиль с водителем

    myCar.Drive(); // Автомобиль начинает движение
    Igor.Drive(myCar); // Водитель также может начать движение, используя автомобиль
```

### Объяснение:

- **Абстрактный класс `Vehicle`:** Определяет базовые свойства и методы для всех транспортных средств. В данном случае, метод `Drive` является абстрактным, что требует его переопределения в производных классах.

- **Класс `Car`:** Наследуется от `Vehicle` и добавляет специфичную для автомобиля логику. Он агрегирует `Driver`, но может существовать и функционировать независимо от него.

- **Класс `Driver`:** Определяет водителя с его атрибутами и методом `Drive`, который позволяет водителю управлять любым транспортным средством, включая автомобиль.

- **Объекты `Igor` и `myCar`:** Создаются экземпляры классов `Driver` и `Car`, демонстрируя агрегацию, где `Car` зависит от `Driver` для своего функционирования, но `Driver` может существовать и функционировать независимо от `Car`.

Этот пример демонстрирует агрегацию, показывая, как классы могут взаимодействовать и зависеть друг от друга без формирования строгих отношений наследования или композиции.

<h4 style="color:DodgerBlue">Заключение</h4>

Основное различие между композицией и агрегацией заключается в степени зависимости между классами:

* Композиция подразумевает тесную связь, где одна часть не может существовать без другой.
* Агрегация означает более слабую связь, где части могут существовать и функционировать независимо.

Выбор между композицией и агрегацией зависит от конкретных требований к системе и степень зависимости между компонентами.

Ниже в блоке по примеру создайте класс Animal и реализуйте агрегацию 

````polyglot-notebook
class Owner
{
    public string Name { get; set; }

    // Конструктор вызывается при new Owner(...).
    public Owner(string name)
    {
        Name = name;
    }

    public void Introduce()
    {
        Console.WriteLine($"Хозяин: {Name}");
    }
}

class Animal
{
    private readonly Owner owner;
    public string Name { get; set; }

    // Принимаем уже созданного хозяина, вместо создания его внутри.
    public Animal(string name, Owner owner)
    {
        Name = name;
        this.owner = owner;
    }

    public void ShowOwner()
    {
        Console.WriteLine($"Животное: {Name}. Хозяин: {owner.Name}.");
    }
}

Owner owner = new Owner("Дмитрий");
Animal aggregatedDog = new Animal("Шарик", owner);
Animal aggregatedCat = new Animal("Мурка", owner);

aggregatedDog.ShowOwner();
aggregatedCat.ShowOwner();
owner.Introduce();
````

### Разбор: агрегация

**Теория.** Агрегация — более слабая связь между объектами, которые могут существовать независимо. В примере хозяин создаётся отдельно от животных и может быть связан сразу с несколькими из них. Для композиции и агрегации в C# нет специальных ключевых слов: различие задаёт устройство модели и владение объектами.

**Что происходит в решении:**
1. `new Owner("Дмитрий")` создаёт хозяина и вызывает конструктор Owner. Конструктор имеет имя класса и подготавливает объект.
2. `new Animal("Шарик", owner)` создаёт животное и передаёт ссылку на уже существующего хозяина.
3. В `this.owner = owner` слева находится поле текущего животного, справа — параметр конструктора. `this` означает текущий объект.
4. Аналогично создаётся Мурка; оба животных ссылаются на одного хозяина, а не на две его копии.
5. `ShowOwner()` выводит имена животных и хозяина. `owner.Introduce()` можно вызвать отдельно от животных.

Вывод: `Животное: Шарик. Хозяин: Дмитрий.`, `Животное: Мурка. Хозяин: Дмитрий.`, `Хозяин: Дмитрий`.
