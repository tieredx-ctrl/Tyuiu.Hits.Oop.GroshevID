### Структура и Объявление Класса в C#

В C#, класс объявляется с использованием ключевого слова `class`, за которым следует имя класса. Класс может содержать поля (данные), свойства (способ доступа к данным), методы (функциональность) и события. 

#### Свойства

Свойства в C# используются для инкапсуляции данных. Они предоставляют гибкий механизм для чтения, записи или вычисления значений приватных полей. Свойства могут иметь два акцессора: `get` (геттер) для чтения значения и `set` (сеттер) для его изменения.

#### Геттеры и Сеттеры

Геттеры и сеттеры являются частью свойств класса в C#. Геттер (`get`) используется для возврата значения свойства, а сеттер (`set`) — для установки нового значения. Сеттер может включать логику валидации или другие операции перед изменением значения поля.

### Пример Классов Vehicle и Car

Давайте рассмотрим примеры классов `Vehicle` и `Car`, чтобы проиллюстрировать использование свойств, геттеров и сеттеров, а также разницу между статическими и обычными атрибутами и методами.

#### Класс Vehicle

```csharp
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}
```

#### Класс Car

```csharp
public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // Задаем начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}
```

### Разница между Статическими и Обычными Атрибутами и Методами Класса

- **Статические атрибуты и методы** принадлежат самому классу, а не его экземплярам. Они могут быть вызваны без создания объекта класса. В примере выше `TotalVehicles` является статическим атрибутом, а `DisplayTotalVehicles()` — статическим методом класса `Vehicle`.
- **Обычные атрибуты и методы** принадлежат экземплярам класса и могут иметь разные значения для разных объектов. В примере выше `_color`, `_speed` и `_fuelLevel` являются обычными атрибутами, а методы `Accelerate()` и `Refuel()` — обычными методами.

### Свойства, get и set простыми словами

Поле хранит данные внутри объекта. Свойство предоставляет способ читать или менять эти данные. В решении `private int age` — закрытое поле, а `public int Age` — доступное снаружи свойство.

```csharp
public int Age
{
    get { return age; }
    set
    {
        if (value >= 0)
            age = value;
        else
            Console.WriteLine("Возраст не может быть отрицательным.");
    }
}
```

- При `dog.Age = 3` выполняется set. Специальное слово value содержит присваиваемое значение — здесь 3.
- При чтении `dog.Age` выполняется get. Он возвращает значение закрытого поля age.
- При `dog.Age = -1` set проверяет value. Отрицательное значение не записывается; прежний возраст сохраняется.

Age и age — разные имена: C# различает большие и маленькие буквы. Обычно свойства называют с большой буквы, а поля — с маленькой. Проверка setter выполняется и при изменении свойства из метода объекта.

**Автоматическое свойство** можно записать коротко, если дополнительная логика не требуется:

```csharp
public string Breed { get; set; } = "Не указана";
```

Компилятор сам создаёт скрытое поле для хранения значения. В нашем примере это подходит для породы собаки и цвета кошки. Для Age записан полный setter с проверкой.

**Разница обычных и статических свойств:** имя и возраст принадлежат отдельному животному. Count — общее свойство класса: оно считает созданных животных. При new Dog() или new Cat() вызывается конструктор Animal, который увеличивает Count.

```csharp
public static int Count { get; private set; } = 0;
```

Get публичен, поэтому число можно прочитать извне. Private set позволяет менять его внутри Animal. Метод ShowCount вызывается через класс: Animal.ShowCount(). Обычный HaveBirthday вызывается через объект: dog.HaveBirthday().

Конструкция `Age++` читает свойство, увеличивает значение на 1 и записывает его через setter. Нельзя присвоить значение непосредственно private-полю age из основного кода.

В исходной теории для ошибочного значения показано исключение. В этих учебных примерах setter выводит понятное сообщение и сохраняет прежнее значение, чтобы весь пример продолжил выполнение. Примеры сосредоточены на свойствах и методах текущего таска.

````polyglot-notebook
public class Vehicle
{
    private string color = "Белый";
    private int speed = 0;

    public string Color
    {
        get { return color; }
        set { color = value; }
    }

    public int Speed
    {
        get { return speed; }
        set
        {
            if (value >= 0)
                speed = value;
            else
                Console.WriteLine("Скорость не может быть отрицательной.");
        }
    }

    public static int Count { get; private set; } = 0;

    public Vehicle()
    {
        Count++;
    }

    public void Accelerate(int amount)
    {
        Speed += amount;
    }

    public static void ShowCount()
    {
        Console.WriteLine($"Создано транспортных средств: {Count}.");
    }
}

public class Car : Vehicle
{
    public int Fuel { get; set; } = 0;

    public void Refuel(int amount)
    {
        Fuel += amount;
    }
}

Car firstCar = new Car();
firstCar.Color = "Зелёный";
firstCar.Speed = 20;
firstCar.Fuel = 5;
Car secondCar = new Car();
secondCar.Color = "Белый";
secondCar.Fuel = 10;

firstCar.Accelerate(10);
firstCar.Refuel(5);
Console.WriteLine($"Первая машина: {firstCar.Color}, скорость {firstCar.Speed}, топливо {firstCar.Fuel}.");
Console.WriteLine($"Вторая машина: {secondCar.Color}, скорость {secondCar.Speed}, топливо {secondCar.Fuel}.");
firstCar.Speed = -5;
Console.WriteLine($"Скорость после попытки записать -5: {firstCar.Speed}.");
Vehicle.ShowCount();

````

````polyglot-notebook
public class Animal
{
    private string name = "Без имени";
    private int age = 0;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public int Age
    {
        get { return age; }
        set
        {
            if (value >= 0)
                age = value;
            else
                Console.WriteLine("Возраст не может быть отрицательным.");
        }
    }

    public static int Count { get; private set; } = 0;

    public Animal()
    {
        Count++;
    }

    public void HaveBirthday()
    {
        Age++;
    }

    public static void ShowCount()
    {
        Console.WriteLine($"Создано животных: {Count}.");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; } = "Не указана";

    public void Bark()
    {
        Console.WriteLine($"{Name} лает.");
    }
}

public class Cat : Animal
{
    public string Color { get; set; } = "Не указан";

    public void Meow()
    {
        Console.WriteLine($"{Name} мяукает.");
    }
}

Dog dog = new Dog();
dog.Name = "Шарик";
dog.Age = 3;
dog.Breed = "Такса";
Cat cat = new Cat();
cat.Name = "Мурка";
cat.Age = 2;
cat.Color = "Рыжий";

dog.HaveBirthday();
Console.WriteLine($"Собака: {dog.Name}, возраст: {dog.Age}, порода: {dog.Breed}.");
Console.WriteLine($"Кошка: {cat.Name}, возраст: {cat.Age}, цвет: {cat.Color}.");
dog.Bark();
cat.Meow();
dog.Age = -1;
Console.WriteLine($"Возраст собаки после попытки записать -1: {dog.Age}.");
Animal.ShowCount();

````
